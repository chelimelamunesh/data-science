```python
#what is regressions 
##{answer: Regression is a statistical method used to find the relationship between independent variables (X) and dependent variable (Y). It helps in prediction and forecasting.
#types of regressions :
# 1️⃣ Linear Regression → Predicts continuous values (e.g., house prices, sales).
Y=mX+c
Example: Predicting salary based on experience.
# 2️⃣ Logistic Regression → Used for classification (Yes/No, 0/1).

P(Y) = 1/1+e^-(b0+b1X)
Example: Predicting if a customer will buy a product (Yes/No).
# 3️⃣ Multiple Regression → Uses multiple X variables to predict Y.

Y = b0+ b1X1 + b2X2+......+bnXn
 
Example: Predicting house price based on area, location, and number of rooms
#use in data science
✔️ Prediction (e.g., stock prices, sales forecasting).
✔️ Finding Relationships (e.g., how temperature affects ice cream sales).
✔️ Feature Selection (choosing important variables).
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # Independent variable
Y = np.array([2, 4, 6, 8, 10])  # Dependent variable

model = LinearRegression()
model.fit(X, Y)

print(f"Prediction for X=6: {model.predict([[6]])[0]}")
```

```python
#why regression
Regression is one of the most important techniques in data science, statistics, and machine learning because it helps in:

1️⃣ Prediction & Forecasting
Example: Predict house prices based on size, location, and number of rooms.
Use Case: Sales forecasting, weather prediction, stock market trends.

2️⃣ Finding Relationships Between Variables
Example: Does advertising spend increase sales?
Use Case: Marketing analysis, business decision-making.

3️⃣ Feature Selection & Impact Analysis
Helps in understanding which variables matter the most.
Example: In predicting student performance, is study time or sleep more important?
Use Case: Healthcare (which symptoms strongly indicate a disease?).

4️⃣ Anomaly Detection
Regression helps find outliers (data points that don’t fit the pattern).
Example: Detecting fraudulent transactions in banking.

5️⃣ Optimization & Decision Making
Used in resource allocation, risk management, and pricing strategies.
Example: Setting the best price for a product based on demand and production costs.

Conclusion: Why Regression?
✔ Easy to interpret results.
✔ Powerful for prediction & decision-making.
✔ Used in almost every industry (finance, healthcare, marketing, tech).
```

```python
from sklearn.linear_model import LinearRegression
import numpy as np

# Independent variable (House size in square feet)
X = np.array([1000, 1500, 2000, 2500, 3000]).reshape(-1, 1)

# Dependent variable (House Price in lakhs)
Y = np.array([50, 75, 100, 125, 150])

# Train the model
model = LinearRegression()
model.fit(X, Y)

# Predict price for a house of 1800 sq ft
prediction = model.predict([[1800]])
print(f"Predicted Price: {prediction[0]} Lakhs")
```

```python
# root mean square error
rmse_lin = np.sqrt(np.mean((np.array(Data['AT'])-np.array(pred))**2))
rmse_lin
```

```python
plt.scatter(x = Data['Waist'], y = Data["AT"],color = 'red')
plt.plot(Data['Waist'],pred,color = 'black')
plt.xlabel('Waist')
plt.ylabel('isuue')
```

```python
rmse1 = np.sqrt(np.mean((np.array(Data['AT'])-np.array(pred2))**2))
rmse1
```

```python
#exponential model
model3 = smf.ols('np.log(AT)~ Waist',data = Data).fit()
model3
```

```python
model3.summary()
```

```python
pred3 = model3.predict(pd.DataFrame(Data['Waist']))
print(pred3)
```

```python
#as we have used log(AT) in preparing model so we need to convert it back
np.exp(pred3)
```

```python
rmse_exp = np.sqrt(np.mean((np.array(Data['AT'])-np.array(pred3))**2))
rmse_exp
```
