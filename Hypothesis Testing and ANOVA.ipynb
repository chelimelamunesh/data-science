```python
from statsmodels.stats import proportion 

n1, n2 = 247, 308
p1, p2 = 0.368, 0.389
x1, x2 = p1*n1, p2*n2

test_statistic, p_value = proportion.proportions_ztest([x1,x2], [n1,n2], alternative = 'two-sided')
print(test_statistic, p_value)
#alternative='two-sided' → Specifies that the test is two-tailed (checking for any significant difference, whether greater or smaller).
alpha = 0.05

if p_value < alpha:
    print("we reject the H0")
else: 
    print("we fail to reject the H0")
```

```python
# Use 'two-sided' when checking for any difference (p1 ≠ p2)
# Use 'larger' when testing if p1 > p2
# Use 'smaller' when testing if p1 < p2
```
