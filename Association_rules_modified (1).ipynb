```python
# Import Data from CSV file 
import numpy as np
import pandas as pd 
dataset = pd.read_csv( r"C:\Users\munes\Dropbox\PC\Documents\assignments\assignments\groceries - groceries.csv",sep=',')
dataset = dataset.drop(['Item(s)'], axis=1)
dataset = dataset.replace(np.nan, '', regex=True)
dataset
```

```python
dataset
transactions = []
for i in range(0, 9835):
    transactions.append([str(dataset.values[i,u]) for u in range(1, 31)]) 
```

```python
groceries_series  = pd.DataFrame(pd.Series(transactions))
groceries_series 
groceries_series.columns = ["business"]
groceries_series
```

```python
X = groceries_series['business'].str.join(sep=',').str.get_dummies(sep=',')
X
'''✅ sep=',' is used to split multiple values in a single cell before one-hot encoding.
✅ Without sep=',', multi-category values would be treated as a single category.
✅ Use sep=',' when dealing with multiple values stored as comma-separated strings in a column.'''
```

```python
import warnings
warnings.filterwarnings('ignore')
from mlxtend.frequent_patterns import apriori,association_rules
frequent_itemsets = apriori(X, min_support=0.005,max_len=3,use_colnames = True)
frequent_itemsets 
```

```python
frequent_itemsets.sort_values('support',ascending = False,inplace=True)
frequent_itemsets.sort_values 
```

```python
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1)
rules.head(20)
rules.sort_values('lift',ascending = False).head(10)  
```

```python
'''Association Rule Mining: A Complete Guide
What is Association Rule Mining?
Association rule mining is a data mining technique used to discover relationships between items in large datasets. It is primarily used in market basket analysis, where we analyze customer purchasing behavior to find patterns like:

"If a customer buys bread, they are likely to buy butter."
"If a person buys a laptop, they may also buy a mouse."
This helps businesses make data-driven decisions, such as: ✅ Placing frequently bought-together items close to each other in stores.
✅ Offering discounts on associated products.
✅ Improving recommendation systems (like Amazon, Flipkart, etc.).

Where Do We Use Association Rule Mining?
🔹 Retail and E-commerce: To analyze customer purchasing behavior.
🔹 Healthcare: Finding co-occurrence of diseases or medication patterns.
🔹 Finance: Detecting fraudulent transactions.
🔹 Web Usage Mining: Understanding user navigation on websites.
🔹 Recommendation Systems: Netflix, Amazon, and YouTube use this to suggest relevant items.

Steps in Association Rule Mining
The most commonly used algorithm for association rule mining is Apriori Algorithm.
Let’s go step by step:

Step 1: Data Collection
Gather transaction data, where each row represents a transaction and each column represents an item.
Example dataset:
Transaction ID	Milk	Bread	Butter	Eggs
1	✅	✅	❌	✅
2	✅	✅	✅	❌
3	❌	✅	✅	✅
4	✅	❌	✅	✅
👉 Goal: Find patterns like "People who buy Bread also buy Butter."

Step 2: Convert Data into a Transaction Format
We need data in list format, where each transaction contains only the items that were purchased.
 
transactions = [
    ['Milk', 'Bread', 'Eggs'],
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter', 'Eggs'],
    ['Milk', 'Butter', 'Eggs']
]
👉 This makes it easier to apply the Apriori algorithm.

Step 3: Find Frequent Itemsets
Support measures how frequently an itemset appears in the dataset.

Support
=
Transactions containing item A and B
Total Transactions
 
 
Example:

If Milk appears in 3 out of 4 transactions, its support is 3/4 = 0.75.
We filter frequent itemsets that have support above a threshold (e.g., 0.5).
Step 4: Generate Association Rules
Once frequent itemsets are found, we generate rules in the format:

If (A), then (B)
If (A), then (B)
Each rule has:

Confidence: Measures how often B appears when A is present.

Confidence
=
Transactions containing both A and B
Transactions containing A

 
Lift: Measures how much more likely B is bought when A is bought.

Lift
=
Confidence of A → B
Expected Probability of B

 
Code for Association Rule Mining (Apriori Algorithm)
We will use Python's mlxtend library to implement Apriori.

Install Dependencies
 
pip install mlxtend
Complete Code for Association Rule Mining
 
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# Sample dataset (transactional format)
transactions = [
    ['Milk', 'Bread', 'Eggs'],
    ['Milk', 'Bread', 'Butter'],
    ['Bread', 'Butter', 'Eggs'],
    ['Milk', 'Butter', 'Eggs']
]

# Convert to DataFrame (one-hot encoding)
df = pd.DataFrame([
    {'Milk': 1, 'Bread': 1, 'Butter': 0, 'Eggs': 1},
    {'Milk': 1, 'Bread': 1, 'Butter': 1, 'Eggs': 0},
    {'Milk': 0, 'Bread': 1, 'Butter': 1, 'Eggs': 1},
    {'Milk': 1, 'Bread': 0, 'Butter': 1, 'Eggs': 1}
])

# Step 1: Find Frequent Itemsets (Support ≥ 0.5)
frequent_itemsets = apriori(df, min_support=0.5, use_colnames=True)

# Step 2: Generate Association Rules (Confidence ≥ 0.7)
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)

# Print results
print("Frequent Itemsets:\n", frequent_itemsets)
print("\nAssociation Rules:\n", rules)
Understanding the Output
Frequent Itemsets (Generated by Apriori)
scss
Copy
Edit
   support     itemsets
0      0.75     (Milk)
1      0.75     (Eggs)
2      0.75     (Bread)
3      0.75     (Butter)
4      0.50     (Milk, Eggs)
5      0.50     (Milk, Butter)
6      0.50     (Bread, Butter)
👉 This shows which items are frequently bought together.

Association Rules
scss
 
antecedents  consequents  support  confidence  lift
(Milk)       → (Eggs)       0.50     0.67       1.33
(Bread)      → (Butter)     0.50     0.67       1.50
👉 This tells us:

If Milk is bought, there’s a 67% chance that Eggs will also be bought.
If Bread is bought, there’s a 67% chance that Butter will also be bought.
Key Takeaways
✅ Association rule mining is used to find hidden patterns in data.
✅ Apriori Algorithm helps identify frequently bought items together.
✅ Metrics like Support, Confidence, and Lift help evaluate rules.
✅ Useful in retail, healthcare, finance, web usage mining, and recommendation systems.'''
```
