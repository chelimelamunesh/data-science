```python
import numpy as np
import scipy.stats as stats

# Sample data
data = [23, 20, 19, 25, 27, 24, 22, 26, 28, 30]

# Step 1: Calculate the sample mean and standard error
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)  # ddof=1 for sample standard deviation
n = len(data)
standard_error = sample_std / np.sqrt(n)

# Step 2: Define the confidence level (e.g., 95%)
confidence_level = 0.95
z_critical = stats.norm.ppf((1 + confidence_level) / 2)  # For large samples

# Step 3: Calculate the margin of error
margin_of_error = z_critical * standard_error

# Step 4: Compute the confidence interval
confidence_interval = (sample_mean - margin_of_error, sample_mean + margin_of_error)

print(f"Sample Mean: {sample_mean}")
print(f"95% Confidence Interval: {confidence_interval}")
```

```python
# STEPS NEED TO FOLLOW IN HYPOTHESES TESTING
#Hypothesis testing is used to make inferences about a population from sample data. Key steps include:
#1.Define the null hypothesis (𝐻0) and alternative hypothesis (𝐻𝑎).
#2.Choose the significance level (𝛼), often 0.05.
#3.Select an appropriate test based on:
    #3.1 The type of data (numerical/categorical).
    #3.2 Number of samples (1-sample, 2-sample, paired data)
#4.Calculate the test statistic and p-value.
#5. Make a decision
  #5.1 If 𝑝≤α, reject H 0
  #5.2 If p>α, fail to reject H0
```

```python
# 1-SAMPLE TEST
#PURPOSE: Tests whether the mean of a single sample differs from a known or hypothesized value.
# EXAMPLE : A company claims the average weight of a product is 50g. You collect a sample and want to test this claim.
import scipy.stats as stats

# Data
sample = [48, 52, 49, 51, 50, 50, 49, 52]
population_mean = 50

# Perform 1-sample t-test
t_stat, p_value = stats.ttest_1samp(sample, population_mean)
print(f"T-Statistic: {t_stat}, P-Value: {p_value}")

# Conclusion
alpha = 0.05
if p_value <= alpha:
    print("Reject the null hypothesis. The sample mean is significantly different.")
else:
    print("Fail to reject the null hypothesis. The sample mean is not significantly different.")
```

```python
# 3. 2-Sample t-Test
# Purpose:Tests whether the means of two independent samples are significantly different.
# EXAMPLE : Compare the average test scores of two student groups.
# Data
group1 = [85, 88, 89, 91, 87]
group2 = [80, 78, 84, 79, 77]

# Perform 2-sample t-test
t_stat, p_value = stats.ttest_ind(group1, group2)
print(f"T-Statistic: {t_stat}, P-Value: {p_value}")

# Conclusion
if p_value <= alpha:
    print("Reject the null hypothesis. The group means are significantly different.")
else:
    print("Fail to reject the null hypothesis. The group means are not significantly different.")
```

```python
#5. Paired t-Test
# PURPOSE: Tests whether the means of paired observations differ (e.g., before-and-after measurements).
# EXAMPLE : Compare students' scores before and after training.
# Data
before = [85, 87, 86, 84, 88]
after = [90, 89, 88, 91, 92]

# Perform paired t-test
t_stat, p_value = stats.ttest_rel(before, after)
print(f"T-Statistic: {t_stat}, P-Value: {p_value}")

if p_value <= alpha:
    print("Reject the null hypothesis. The means are significantly different.")
else:
    print("Fail to reject the null hypothesis.")
```

```python
#6. Mann-Whitney U Test
#PURPOSE : Tests whether two independent samples come from the same distribution (non-parametric alternative to t-test). 
# EXAMPLE : Compare customer satisfaction scores of two groups.
# Data
group1 = [3, 5, 4, 6, 4]
group2 = [7, 8, 9, 7, 6]

# Perform Mann-Whitney U Test
u_stat, p_value = stats.mannwhitneyu(group1, group2)
print(f"U-Statistic: {u_stat}, P-Value: {p_value}")
```

```python
#7. One-Way ANOVA (Analysis of Variance)
# PURPOSE : Tests whether the means of three or more groups are significantly different.
# EXAMPLE : Compare test scores across three classes.
# Data
group1 = [85, 86, 88, 85]
group2 = [80, 82, 81, 83]
group3 = [90, 91, 89, 92]

# Perform One-Way ANOVA
f_stat, p_value = stats.f_oneway(group1, group2, group3)
print(f"F-Statistic: {f_stat}, P-Value: {p_value}")

if p_value <= alpha:
    print("Reject the null hypothesis. At least one group mean is different.")
else:
    print("Fail to reject the null hypothesis.")
```

```python
#9. Kruskal-Wallis Test
# PURPOSE : Tests whether the medians of three or more groups are different (non-parametric alternative to ANOVA).
# EXAMPLE : Compare performance scores across three teams.
# Data
group1 = [5, 7, 8, 6, 9]
group2 = [10, 12, 11, 9, 13]
group3 = [15, 14, 16, 17, 15]

# Perform Kruskal-Wallis Test
h_stat, p_value = stats.kruskal(group1, group2, group3)
print(f"H-Statistic: {h_stat}, P-Value: {p_value}")
```

```python
#10. 1-Proportion Test
#PURPOSE : Tests whether a population proportion differs from a hypothesized value.

from statsmodels.stats.proportion import proportions_ztest

# Data
successes = 40
n = 100
p_null = 0.5

# Perform 1-proportion test
z_stat, p_value = proportions_ztest(successes, n, value=p_null)
print(f"Z-Statistic: {z_stat}, P-Value: {p_value}")
```

```python
#11. 2-Proportion Test
# PURPOSE : Tests whether two population proportions are significantly different.

# Data
successes = [40, 50]
n = [100, 120]

# Perform 2-proportion test
z_stat, p_value = proportions_ztest(successes, n)
print(f"Z-Statistic: {z_stat}, P-Value: {p_value}")
```

```python
#12. Chi-Square Test
# PURPOSE : Tests relationships between categorical variables.
# EXAMPLE : Does gender influence product preference?

import pandas as pd
from scipy.stats import chi2_contingency

# Data
data = [[50, 30], [20, 40]]  # Rows: Gender (Male, Female); Columns: Preference (A, B)
chi2, p_value, dof, expected = chi2_contingency(data)
print(f"Chi-Square: {chi2}, P-Value: {p_value}, Degrees of Freedom: {dof}")
```
