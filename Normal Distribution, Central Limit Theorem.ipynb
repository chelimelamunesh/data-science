```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data into a DataFrame
# beml_df = pd.read_csv('your_file.csv')  # Uncomment if loading from a file
print(beml_df.head())  # Check the first few rows

# Ensure 'Date' is a datetime type
beml_df['Date'] = pd.to_datetime(beml_df['Date'])
beml_df.set_index('Date', inplace=True)
```

```python
plt.figure(figsize=(12, 6))
plt.plot(beml_df['Close'], label='Close Price', color='blue')
plt.title('BEML Stock Close Price Over Time', fontsize=16)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Price', fontsize=12)
plt.legend()
plt.grid()
plt.show()
```

```python
# Calculate daily percentage change
beml_df['Daily Change %'] = beml_df['Close'].pct_change() * 100

# Plot daily percentage change
plt.figure(figsize=(12, 6))
plt.plot(beml_df['Daily Change %'], label='Daily Change %', color='orange')
plt.axhline(0, color='red', linestyle='--', linewidth=1)  # Mark zero line
plt.title('BEML Daily Percentage Change', fontsize=16)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Daily Change %', fontsize=12)
plt.legend()
plt.grid()
plt.show()
```

```python
# Calculate moving averages
beml_df['SMA_50'] = beml_df['Close'].rolling(window=50).mean()
beml_df['SMA_200'] = beml_df['Close'].rolling(window=200).mean()

# Plot Close price with moving averages
plt.figure(figsize=(12, 6))
plt.plot(beml_df['Close'], label='Close Price', color='blue')
plt.plot(beml_df['SMA_50'], label='50-Day SMA', color='green', linestyle='--')
plt.plot(beml_df['SMA_200'], label='200-Day SMA', color='red', linestyle='--')
plt.title('BEML Close Price with Moving Averages', fontsize=16)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Price', fontsize=12)
plt.legend()
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.plot(beml_df['High'], label='High Prices', color='green')
plt.plot(beml_df['Low'], label='Low Prices', color='red')
plt.title('BEML High and Low Prices Over Time', fontsize=16)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Price', fontsize=12)
plt.legend()
plt.grid()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.bar(beml_df.index, beml_df['Total Trade Quantity'], color='orange', label='Trade Volume')
plt.title('BEML Trade Volume Over Time', fontsize=16)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Trade Volume', fontsize=12)
plt.legend()
plt.grid()
plt.show()
```

```python
def calculate_rsi(data, window=14):
    delta = data['Close'].diff(1)
    gain = delta.where(delta > 0, 0)
    loss = -delta.where(delta < 0, 0)

    avg_gain = gain.rolling(window=window, min_periods=1).mean()
    avg_loss = loss.rolling(window=window, min_periods=1).mean()

    rs = avg_gain / avg_loss
    rsi = 100 - (100 / (1 + rs))
    return rsi

beml_df['RSI'] = calculate_rsi(beml_df, window=14)

plt.figure(figsize=(12, 6))
plt.plot(beml_df['RSI'], label='RSI', color='purple')
plt.axhline(70, color='red', linestyle='--', linewidth=1, label='Overbought')
plt.axhline(30, color='green', linestyle='--', linewidth=1, label='Oversold')
plt.title('BEML RSI', fontsize=16)
plt.xlabel('Date', fontsize=12)
plt.ylabel('RSI', fontsize=12)
plt.legend()
plt.grid()
plt.show()
```

```python
# Calculate Bollinger Bands
beml_df['SMA_20'] = beml_df['Close'].rolling(window=20).mean()
beml_df['Upper Band'] = beml_df['SMA_20'] + (beml_df['Close'].rolling(window=20).std() * 2)
beml_df['Lower Band'] = beml_df['SMA_20'] - (beml_df['Close'].rolling(window=20).std() * 2)

plt.figure(figsize=(12, 6))
plt.plot(beml_df['Close'], label='Close Price', color='blue')
plt.plot(beml_df['Upper Band'], label='Upper Band', color='red', linestyle='--')
plt.plot(beml_df['Lower Band'], label='Lower Band', color='green', linestyle='--')
plt.fill_between(beml_df.index, beml_df['Lower Band'], beml_df['Upper Band'], color='gray', alpha=0.1)
plt.title('BEML Bollinger Bands', fontsize=16)
plt.xlabel('Date', fontsize=12)
plt.ylabel('Price', fontsize=12)
plt.legend()
plt.grid()
plt.show()
```
