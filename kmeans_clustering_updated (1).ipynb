```python
def norm_func(i):
    x = (i-i.min()) / (i.max() - i.min())
    return (x)  
```

```python
df_norm = norm_func(Univ.iloc[:,1:])  
```

```python
'''Explanation:
WCSS = []: Initializes an empty list to store Within-Cluster Sum of Squares (WCSS) values.
for i in range(1, 11): Iterates over cluster numbers from 1 to 10.
clf = KMeans(n_clusters=i): Creates a K-Means model with i clusters.
clf.fit(df_norm): Trains the model on normalized data (df_norm).
WCSS.append(clf.inertia_): Stores the inertia (WCSS) value for each cluster count.
plt.plot(range(1, 11), WCSS): Plots the WCSS values to find the "elbow point", where WCSS stops decreasing significantly.
Purpose:
The elbow point in the graph suggests the optimal number of clusters, helping in choosing the best K for K-Means.'''
```

```python
from sklearn.cluster import KMeans
fig = plt.figure(figsize=(10, 8))
WCSS = []
for i in range(1, 11):
    clf = KMeans(n_clusters=i)
    clf.fit(df_norm)
    WCSS.append(clf.inertia_) '''# inertia is another name for WCSS,
    The append() method in Python is used to add an element to the end of a list. It modifies the list in place and does not return a new list.'''


plt.plot(range(1, 11), WCSS)
plt.title('The Elbow Method')
plt.ylabel('WCSS')
plt.xlabel('Number of Clusters')
plt.show() 
```

```python
'''Drawbacks of K-Means Clustering
1️⃣ Needs Predefined K (Number of Clusters)

You must specify the number of clusters (k) beforehand, which is often unknown.
Choosing an incorrect k can lead to poor clustering results.
2️⃣ Sensitive to Initial Centroids

Different initializations can lead to different clustering outcomes.
Poor initialization may result in local optima rather than the best clusters.
3️⃣ Not Suitable for Non-Spherical Clusters

K-Means assumes clusters are spherical and evenly sized, which may not always be true.
It struggles with clusters of irregular shapes or varying densities.
4️⃣ Sensitive to Outliers

Outliers can distort centroids, leading to incorrect clustering.
A few extreme values can pull the cluster centers away from their ideal positions.
5️⃣ Does Not Work Well with Categorical Data

K-Means relies on Euclidean distance, which does not work well for categorical variables.
For mixed data (numerical + categorical), K-Modes or K-Prototypes are better alternatives.
6️⃣ Assumes Equal Cluster Sizes

If clusters have different densities or sizes, K-Means tends to assign more points to larger clusters.
It does not perform well when some clusters are much denser or smaller than others.
7️⃣ Fails with Overlapping Clusters

If clusters overlap, K-Means may misclassify points, leading to poor cluster separation.
This happens because it relies on distance-based partitioning.
8️⃣ High Dimensionality Affects Performance

In high-dimensional data, Euclidean distance becomes less meaningful, reducing K-Means effectiveness.
Dimensionality reduction (like PCA) is often needed before applying K-Means.
9️⃣ Computational Complexity for Large Data

With large datasets, K-Means can be computationally expensive due to repeated distance calculations.
The complexity is O(n * k * d), where n = samples, k = clusters, d = dimensions.
🔟 Does Not Guarantee the Best Solution

Since K-Means may converge to a local minimum, different runs can produce different results.
Using K-Means++ initialization helps mitigate this issue.

✅ Best for: Simple, well-separated clusters with numerical data.
❌ Not ideal for: Non-spherical, overlapping, high-dimensional, or categorical data.'''
```

```python
'''Steps to Perform K-Means Clustering
1️⃣ Choose the Number of Clusters (K)
Determine the number of clusters (K) you want to create.
Use the Elbow Method or Silhouette Score to find the optimal K.
2️⃣ Initialize Cluster Centroids
Select K random points from the dataset as initial centroids.
Use K-Means++ initialization for better results.
3️⃣ Assign Data Points to the Nearest Cluster
Calculate the Euclidean distance between each data point and the centroids.
Assign each point to the nearest centroid (forming clusters).
4️⃣ Compute New Centroids
Calculate the mean of all points in each cluster.
Update the centroids to these new mean values.
5️⃣ Repeat Steps 3 and 4 (Iterate Until Convergence)
Reassign points to the closest centroid.
Recalculate new centroids.
Repeat until the centroids no longer change (or a stopping condition is met).
6️⃣ Evaluate the Model
Check cluster quality using WCSS (Within-Cluster Sum of Squares).
Use Silhouette Score to measure how well-separated the clusters are.
7️⃣ Visualize the Clusters (Optional)
If data is 2D or 3D, use scatter plots to visualize clusters.
For higher dimensions, use PCA (Principal Component Analysis) to reduce dimensions and plot.
8️⃣ Make Predictions for New Data
Once trained, use the K-Means model to assign new data points to existing clusters.'''
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Sample Data
data = {'X': [2, 3, 10, 12, 30, 40, 50],
        'Y': [4, 5, 15, 18, 50, 60, 80]}
df = pd.DataFrame(data)

# Apply K-Means with K=2
kmeans = KMeans(n_clusters=2, random_state=42)
df['Cluster'] = kmeans.fit_predict(df)

# Plot Clusters
plt.scatter(df['X'], df['Y'], c=df['Cluster'], cmap='coolwarm', s=100)
plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], 
            marker='X', s=200, c='black', label='Centroids')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()
plt.show()
```
