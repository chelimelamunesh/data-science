```python
import numpy as np
import pandas as pd
import matplotlib.pylab as plt 
Univ = pd.read_csv(r"C:\Users\munes\Dropbox\PC\Documents\assignments\assignments\Universities_Clustering.csv")
Univ    
```

```python
def norm_func(i):
    x = (i-i.min())/(i.max()-i.min())
    return (x)  
```

```python
df_norm = norm_func(Univ.iloc[:,1:])
df_norm
  
```

```python
from sklearn.preprocessing import MinMaxScaler
trans = MinMaxScaler()
data = pd.DataFrame(trans.fit_transform(Univ.iloc[:,1:]))
data   
```

```python
from scipy.cluster.hierarchy import linkage 
import scipy.cluster.hierarchy as sch # for creating dendrogram 
#p = np.array(df_norm) # converting into numpy array format 
z = linkage(df_norm, method="average",metric="euclidean")
plt.figure(figsize=(15, 5))
plt.title('Hierarchical Clustering Dendrogram')
plt.xlabel('Index')
plt.ylabel('Distance')
sch.dendrogram(
    z,
    #leaf_rotation=6.,  # rotates the x axis labels
    #leaf_font_size=15.,  # font size for the x axis labels
)
plt.show()     
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

# Sample data points
X = np.array([[2, 3], [3, 4], [5, 6], [8, 8], [7, 9], [4, 2]])

# Applying K-Means with 2 clusters
kmeans = KMeans(n_clusters=2, random_state=42)
kmeans.fit(X)

# Getting cluster centroids and labels
centroids = kmeans.cluster_centers_
labels = kmeans.labels_

# Plot the clusters
plt.scatter(X[:, 0], X[:, 1], c=labels, cmap='rainbow', marker='o', label='Data Points')
plt.scatter(centroids[:, 0], centroids[:, 1], c='black', marker='X', s=200, label='Centroids')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('K-Means Clustering')
plt.legend()
plt.show()
```

```python
'''Advantages & Drawbacks of Hierarchical Clustering
✅ Advantages of Hierarchical Clustering
1️⃣ No Need to Specify the Number of Clusters (K)

Unlike K-Means, Hierarchical Clustering does not require predefining K (number of clusters).
You can decide the number of clusters by cutting the dendrogram at the desired level.
2️⃣ Produces a Dendrogram (Tree-Like Structure)

A dendrogram provides a visual representation of the clustering process.
Helps understand the relationships between clusters.
3️⃣ Works Well for Small Datasets

Performs well on small datasets with clear hierarchical structures.
No need for multiple runs like K-Means.
4️⃣ Does Not Depend on Initialization

Unlike K-Means, hierarchical clustering does not depend on random centroid initialization.
Gives consistent results every time.
5️⃣ Captures Nested Cluster Relationships

Identifies clusters within clusters, making it useful for hierarchical structures (e.g., taxonomy, gene analysis).
❌ Drawbacks of Hierarchical Clustering
1️⃣ Computationally Expensive (O(n^2))

It has a high time complexity of O(n^2 log n) for n data points.
Not suitable for large datasets (e.g., millions of points).
2️⃣ Cannot Handle Large Datasets Efficiently

Memory-intensive because it stores distance matrices for all pairs of points.
Becomes impractical for big data.
3️⃣ No Way to Correct Mistakes

Once a point is merged into a cluster, it cannot be reassigned later.
Unlike K-Means, which allows reassignments in each iteration.
4️⃣ Sensitive to Noise & Outliers

Outliers can distort hierarchical structures, leading to poor clustering results.
Agglomerative clustering may form unwanted clusters due to noisy data.
5️⃣ Difficult to Interpret for Large Datasets

Dendrograms become complex and unreadable when applied to large datasets.
Hard to determine the best cut-off point for clusters.
🔹 When to Use Hierarchical Clustering?
✅ Best for small to medium datasets with hierarchical relationships.
✅ Useful when you want a clear visual hierarchy (dendrogram).
✅ Works well if clusters are well-separated and not too large.

🚫 Avoid for large datasets due to high computational cost.
🚫 Not ideal for non-hierarchical or overlapping clusters.'''
```

```python
'''🔹 Steps for Hierarchical Clustering (Agglomerative Approach)
Hierarchical clustering is a step-by-step process where data points are grouped into clusters based on their similarity.

Step 1️⃣: Import Required Libraries
Load necessary Python libraries for clustering and visualization.
python
Copy
Edit'''
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import scipy.cluster.hierarchy as sch
from sklearn.cluster import AgglomerativeClustering
'''Step 2️⃣: Load or Create Dataset
Prepare the dataset for clustering.
'''
data = {'X': [2, 3, 10, 12, 30, 40, 50],
        'Y': [4, 5, 15, 18, 50, 60, 80]}
df = pd.DataFrame(data)
'''Step 3️⃣: Visualize Data
Plot the data to understand its distribution.'''
plt.scatter(df['X'], df['Y'], color='blue')
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Data Distribution')
plt.show()
'''Step 4️⃣: Create a Dendrogram
Use a dendrogram to determine the number of clusters.'''
plt.figure(figsize=(8, 5))
sch.dendrogram(sch.linkage(df, method='ward'))
plt.title('Dendrogram')
plt.xlabel('Data Points')
plt.ylabel('Distance')
plt.show()
'''✅ Identify the optimal number of clusters by looking for the largest vertical gap in the dendrogram.

Step 5️⃣: Apply Agglomerative Clustering
Use Agglomerative Clustering to group similar data points.'''

hc = AgglomerativeClustering(n_clusters=2, affinity='euclidean', linkage='ward')
df['Cluster'] = hc.fit_predict(df)
'''🔹 Explanation:

n_clusters=2: Number of clusters (based on dendrogram).
affinity='euclidean': Distance metric used.
linkage='ward': Minimizes variance within clusters.
Step 6️⃣: Visualize the Final Clusters
Plot the clustered data with different colors.
'''
plt.scatter(df['X'], df['Y'], c=df['Cluster'], cmap='coolwarm', s=100)
plt.xlabel('X')
plt.ylabel('Y')
plt.title('Hierarchical Clustering')
plt.show()
'''🔹 Summary of Steps
1️⃣ Import Libraries
2️⃣ Load/Create Data
3️⃣ Visualize Data (Scatter Plot)
4️⃣ Create a Dendrogram (Determine optimal K)
5️⃣ Apply Agglomerative Clustering
6️⃣ Visualize the Final Clusters'''
```
